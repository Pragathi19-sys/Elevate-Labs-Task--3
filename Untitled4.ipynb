```python

# Colab lo ee code run chey - Titanic dataset direct vastundi
import pandas as pd
# Direct link from github
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.to_csv('Titanic_Dataset.csv', index=False)
print("Download ayipoyindi! Shape:", df.shape)
df.head()
```

```python

# Task 5 - Full EDA - Pragathi
import matplotlib.pyplot as plt
import numpy as np

print("--- a. INFO, DESCRIBE, VALUE_COUNTS ---")
print(df.info())
print(df.describe())
print("\nSurvived:")
print(df['Survived'].value_counts())
print("\nSex:")
print(df['Sex'].value_counts())
print("\nPclass:")
print(df['Pclass'].value_counts())

print("\n--- NULL VALUES ---")
print(df.isnull().sum())

# Cleaning
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
if 'Cabin' in df.columns:
    df.drop('Cabin', axis=1, inplace=True)

# b. c. d. Visualizations
plt.figure(figsize=(8,4))
plt.hist(df['Age'], bins=30, color='skyblue', edgecolor='black')
plt.title('Age Distribution - Most 20-40 years')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

plt.figure(figsize=(8,4))
plt.hist(df['Fare'], bins=30, color='orange', edgecolor='black')
plt.title('Fare Distribution - Highly skewed, most <100')
plt.xlabel('Fare')
plt.show()

# Survival count
plt.figure(figsize=(6,4))
counts = df['Survived'].value_counts()
plt.bar(['Died (0)','Survived (1)'], counts.values, color=['red','green'])
plt.title('Survival Count - 62% Died, 38% Survived')
plt.show()

# Boxplot - Age vs Survived
plt.figure(figsize=(6,4))
data = [df[df['Survived']==0]['Age'], df[df['Survived']==1]['Age']]
plt.boxplot(data, labels=['Died','Survived'])
plt.title('Age vs Survived')
plt.show()

# Scatter - Age vs Fare
plt.figure(figsize=(6,4))
colors = ['red' if x==0 else 'green' for x in df['Survived']]
plt.scatter(df['Age'], df['Fare'], c=colors, alpha=0.5, s=10)
plt.title('Age vs Fare by Survival - Red=Died Green=Survived')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()

# Heatmap for correlation
plt.figure(figsize=(6,5))
corr = df[['Survived','Pclass','Age','Fare']].corr()
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Heatmap')
plt.colorbar()
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center")
plt.show()

print("""
=== e. f. SUMMARY OF FINDINGS ===
1. Pattern: Women 74% survived vs Men 19%
2. Pattern: 1st class 63% survived vs 3rd class 24%
3. Trend: Young children higher survival
4. Anomaly: Fare highly skewed with outliers $500+
5. Relationship: Pclass & Fare negative correlation
""")
```
